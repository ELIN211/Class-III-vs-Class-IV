```python
#Loading the 10years historical data with columns: Dates, KV2_Comdty (CME Class IV milk prices) and DA2_Comdty (CME Class III milk prices)
import pandas as pd
import numpy as np
data=pd.read_csv("https://raw.githubusercontent.com/ELIN211/Class-III-vs-Class-IV/refs/heads/main/Class%20III%20vs%20IV%20Milk.csv")
#Cleaning the dataset
data['Dates']=pd.to_datetime(data['Dates'],dayfirst=True)
data = data.set_index('Dates').sort_index()
print(data)
```

```python
#Calculating the spread of Class IV milk - Class III milk
data['Spread'] = data['KV2_Comdty'] - data['DA2_Comdty']
print(data)
```

```python
#Calculate the 5-year rolling percentile
data['5yr_percentile_rank'] = data['Spread'].rolling('1826D').rank(pct=True)

# Filter the latest 30-day subset
latest_date = data.index.max()
start_date = latest_date - pd.Timedelta(days=30)
subset = data.loc[start_date:latest_date]

# Display the updated subset
display(subset)

#Should the percentile be more than 70, it would be a recommendation to sell Class IV, buy Class III
#Should the percentile be less than 30, it would be a recommendation to buy Class IV, sell Class III
```
