```python
#Loading the 10years historical data with columns: Dates, KV2_Comdty (CME Class IV milk prices) and DA2_Comdty (CME Class III milk prices)
import pandas as pd
import numpy as np
data=pd.read_csv("https://raw.githubusercontent.com/ELIN211/Class-III-vs-Class-IV/refs/heads/main/Class%20III%20vs%20IV%20Milk.csv")
#Cleaning the dataset
data['Dates']=pd.to_datetime(data['Dates'],dayfirst=True)
data = data.set_index('Dates').sort_index()
print(data)
```

```python
#Calculating the spread of Class IV milk - Class III milk
data['Spread'] = data['KV2_Comdty'] - data['DA2_Comdty']
print(data)
```
